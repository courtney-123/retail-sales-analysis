```python
import pandas as pd

# Creating individual rows of messy text data
r1 = [' North ', 5000]
r2 = ['south', 7000]
r3 = ['north', 3000]
r4 = [' East', 4000]
r5 = ['south ', 6000]

cols = ['Region', 'Revenue']
retail_df = pd.DataFrame([r1, r2, r3, r4, r5], columns=cols)

# Show the raw table
retail_df
```

```python
# 1. Standardise messy spacing and uppercase text
retail_df['Region'] = retail_df['Region'].str.strip().str.title()

# 2. Group the data by Region to calculate total revenue
summary = retail_df.groupby('Region')['Revenue'].sum().reset_index()

# Show the clean aggregated table
summary
```

```python
%%writefile cleaner.py
import pandas as pd

def clean_retail_data(raw_records, column_names):
    """
    Automates text standardization and multi-variable data grouping.
    """
    df = pd.DataFrame(raw_records, columns=column_names)
    
    # 1. Standardize messy spacing and uppercase text
    df['Region'] = df['Region'].str.strip().str.title()
    
    # 2. Group the data by Region to calculate total revenue
    summary = df.groupby('Region')['Revenue'].sum().reset_index()
    
    return summary
```
